# 6. Crear nuevas variables

## 6.1 Valor bruto

El valor bruto de un pedido se calcula multiplicando cantidad por precio unitario.


````python
df["valor_bruto"] = df["cantidad"] * df["precio_unitario"]

df[["cantidad", "precio_unitario", "valor_bruto"]].head()
````

## 6.2 Valor del descuento

El porcentaje debe convertirse a una proporción dividiéndolo por 100.


````python
df["valor_descuento"] = (
    df["valor_bruto"] * df["descuento_pct"] / 100
)

df[["valor_bruto", "descuento_pct", "valor_descuento"]].head()
````

✏️ **Completa aquí:** crea `valor_neto` restando el descuento al valor bruto.


````python
# COMPLETA AQUÍ
df["valor_neto"] = df["valor_bruto"] - df["valor_descuento"]

df[["valor_bruto", "valor_descuento", "valor_neto"]].head()
````

👀 **Resultado esperado para el primer pedido:** valor bruto y valor neto de `$285.000`, porque no tiene descuento.

Para continuar, la siguiente celda crea la variable completa por si todavía no terminaste el ejercicio.


````python
# Celda de continuidad
df["valor_neto"] = df["valor_bruto"] - df["valor_descuento"]

round(df["valor_neto"].sum(), 0)
````

👀 **Resultado esperado:** `33505500`

🧠 El valor neto total de los 96 pedidos es $33.505.500.

# ⭐ 11. Introducción a datos relacionales

Hasta ahora trabajamos con una tabla plana que contiene información del cliente y del pedido en cada fila.

Una base relacional puede separar esta información:

## Tabla `clientes`

- `cliente_id`: clave primaria.
- `ciudad`.
- `canal_adquisicion`.
- `tipo_cliente`.

## Tabla `pedidos`

- `pedido_id`: clave primaria.
- `cliente_id`: clave foránea.
- datos del pedido.

La relación es:

> Un cliente puede realizar varios pedidos.


````python
clientes = (
    df[[
        "cliente_id",
        "ciudad",
        "canal_adquisicion",
        "tipo_cliente"
    ]]
    .drop_duplicates()
    .reset_index(drop=True)
)

pedidos = df[[
    "pedido_id",
    "cliente_id",
    "fecha_pedido",
    "categoria_producto",
    "cantidad",
    "precio_unitario",
    "descuento_pct",
    "estado_pedido",
    "satisfaccion",
    "valor_neto"
]].copy()

len(clientes), len(pedidos)
````

👀 **Resultado esperado:** `(30, 96)`

La primera tabla tiene un registro por cliente. La segunda tiene un registro por pedido.


# 12. Ejecutar SQL en Google Colab

Utilizaremos SQLite, una base de datos ligera incluida en Python.


````python
import sqlite3

conexion = sqlite3.connect(":memory:")

clientes.to_sql("clientes", conexion, index=False, if_exists="replace")
pedidos.to_sql("pedidos", conexion, index=False, if_exists="replace")
````

📚 Acabamos de crear dos tablas dentro de una base SQLite temporal.

## Anatomía de una consulta

```sql
SELECT columnas
FROM tabla
WHERE condición
ORDER BY columna
LIMIT cantidad;
```

- `SELECT`: columnas que queremos.
- `FROM`: tabla que consultaremos.
- `WHERE`: registros que cumplen la condición.
- `ORDER BY`: orden del resultado.
- `LIMIT`: cantidad máxima de filas.


## 12.1 Primer ejemplo: pedidos de valor alto


````python
consulta_1 = """
SELECT pedido_id, categoria_producto, valor_neto
FROM pedidos
WHERE valor_neto >= 300000
ORDER BY valor_neto DESC
LIMIT 5;
"""

pd.read_sql_query(consulta_1, conexion)
````

👀 **Resultado esperado:** una tabla con cinco pedidos. `PED0032` aparece primero.

## 12.2 Segundo ejemplo: dos condiciones


````python
consulta_2 = """
SELECT pedido_id, categoria_producto, estado_pedido, valor_neto
FROM pedidos
WHERE estado_pedido = 'Entregado'
  AND valor_neto >= 300000
ORDER BY valor_neto DESC
LIMIT 5;
"""

pd.read_sql_query(consulta_2, conexion)
````

📚 En SQL utilizamos `AND` donde Pandas utiliza `&`.

## 12.3 Comparar SQL y Pandas

La consulta equivalente en Pandas sería:


````python
resultado_pandas = (
    pedidos[
        (pedidos["estado_pedido"] == "Entregado") &
        (pedidos["valor_neto"] >= 300000)
    ]
    .sort_values("valor_neto", ascending=False)
    .head(5)
)

resultado_pandas[
    ["pedido_id", "categoria_producto", "estado_pedido", "valor_neto"]
]
````

🧠 Ambos lenguajes devuelven el mismo resultado. Cambia la forma de expresar la consulta.
