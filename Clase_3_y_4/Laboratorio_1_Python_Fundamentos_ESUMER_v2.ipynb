## 7. Convertir una regla de negocio en una variable

## E3. Una gráfica con propósito


````python
ax = df["Ciudad"].value_counts().plot(
    kind="bar",
    color="#22A1D5",
    title="Clientes por ciudad"
)

ax.set_xlabel("Ciudad")
ax.set_ylabel("Clientes")
````

## E4. Leer un error sin detener la ruta principal

El siguiente ejemplo muestra un error frecuente, pero lo capturamos para que la ejecución pueda continuar.


````python
try:
    df["Edaad"]
except KeyError as error:
    print("Tipo de error: KeyError")
    print("Revisa el nombre de la columna:", error)
    print("Columnas disponibles:", df.columns.tolist())
````
